# Compare saved experiments in native Jupyter

Use Lispyn macros to compare a baseline with a candidate run.

Synthetic metric histories demonstrate the workflow; no model was trained.

```python
import hy
hy.macros.require('lispyn.macros', None, target_module_name='compare_saved_runs', assignments=[['compare-runs', 'compare-runs']], prefix='')
from lispyn.reproducibility import write_json
write_json('runs/compare-demo/baseline/history.json', {'train_loss': [1.2, 0.9, 0.7], 'val_loss': [1.25, 0.95, 0.85]})
write_json('runs/compare-demo/candidate/history.json', {'train_loss': [1.1, 0.75, 0.5], 'val_loss': [1.12, 0.77, 0.55]})
paths = ['runs/compare-demo/baseline', 'runs/compare-demo/candidate']
from lispyn.research_comparison import show_run_comparison
comparison = show_run_comparison(paths, metric='loss', split='val', baseline='baseline', save='runs/compare-demo/comparison.md')
assert comparison['baseline'] == 'baseline'
print('Compared saved histories:', len(comparison['runs']))
hy.macros.require('lispyn.macros', None, target_module_name='compare_saved_runs', assignments=[['md', 'md']], prefix='')
```
