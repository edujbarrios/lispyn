# Train / validation / test and k-fold cross-validation

Synthetic PyTorch dataset: every sample is used exactly once as validation

across the 5 folds. Keep the test split separate until final evaluation.

```python
import hy
hy.macros.require('lispyn.macros', None, target_module_name='cross_validation', assignments=[['train-val-test', 'train-val-test'], ['cross-validation-folds', 'cross-validation-folds']], prefix='')
import torch
from lispyn.cross_validation import summarize_cv_scores
labels = [0] * 15 + [1] * 15
x = torch.arange(30).unsqueeze(1)
data = torch.utils.data.TensorDataset(x, torch.tensor(labels))
from lispyn.data import split_dataset_three
partitions = split_dataset_three(data, labels=labels, validation_fraction=0.2, test_fraction=0.2, seed=42)
print('Train / val / test sizes:', [len(part) for part in partitions])
training = partitions[0]
fold_labels = [labels[index] for index in getattr(training, 'indices')]
scores = []
from lispyn.data import k_fold_dataset
for [fold_train, fold_validation] in k_fold_dataset(training, n_splits=5, labels=fold_labels, seed=42):
    scores.append(len(fold_validation) / len(training))
print(summarize_cv_scores(scores, metric='validation_fraction'))
```
