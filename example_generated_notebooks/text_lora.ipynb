## LoRA fine-tuning for text (offline miniature BERT)
No model downloads. Replace BertConfig with from_pretrained(...) for real models.

```python
import hy
import torch
from transformers import BertConfig, BertForSequenceClassification
from lispyn.peft_utils import lora, parameter_counts
from lispyn.training import fit
config = BertConfig(vocab_size=64, hidden_size=32, num_hidden_layers=1, num_attention_heads=4, intermediate_size=64, num_labels=2)
model = BertForSequenceClassification(config)
model = lora(model, ['query', 'value'], rank=4, task_type='SEQ_CLS')
batch = {'input_ids': torch.randint(0, 64, (3, 8)), 'attention_mask': torch.ones((3, 8), dtype=torch.long), 'labels': torch.tensor([0, 1, 0])}
optimizer = torch.optim.AdamW(filter(lambda p: p.requires_grad, model.parameters()), lr=0.001)
print('LoRA params:', parameter_counts(model))
print(fit(model, [batch], optimizer, epochs=1))
```
