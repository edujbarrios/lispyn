# Confusion matrix with Lispyn

Create a reusable classification report using Lisp macros and the turquoise Docusaurus-dark theme. This example uses **synthetic labels**, not model predictions.

```python
# Lispyn: install missing packages in the active Jupyter kernel
from lispyn.dependencies import ensure_packages
ensure_packages('matplotlib>=3.8')
```

```python
import hy
hy.macros.require('lispyn.macros', None, target_module_name='confusion_matrix', assignments=[['defconfusion', 'defconfusion'], ['confusion-matrix', 'confusion-matrix']], prefix='')
import matplotlib
matplotlib.use('Agg')
labels = ['Cat', 'Dog', 'Bird', 'Fox']
actual = [labels[i % 4] for i in range(60)]
predicted = [labels[(i + 1) % 4 if i % 7 == 0 else i % 4] for i in range(60)]
from lispyn.classification import make_confusion_plotter
dark_classification = make_confusion_plotter(theme='docusaurus-dark', palette='turquoise-spectrum', figsize=[8, 6], title_size=17, dpi=170)
dark_classification(actual, predicted, classes=labels, normalize='true', title='Classification overview', subtitle='Synthetic labels  ·  Docusaurus-dark  ·  Turquoise', save='../docs/images/synthetic_confusion_matrix.png')
print('Saved synthetic confusion-matrix figure.')
hy.macros.require('lispyn.macros', None, target_module_name='confusion_matrix', assignments=[['classification-report', 'classification-report']], prefix='')
import lispyn.classification
scores = lispyn.classification.classification_report(actual, predicted, classes=labels, save='../docs/reports/synthetic_classification_report.md')
print('Classification accuracy:', scores['accuracy'])
```
