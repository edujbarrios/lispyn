## TorchAudio features on a synthetic waveform
No media files, downloads, or GPU needed.

```python
import hy
import torch
import torchaudio
rate = 16000
sample = torch.arange(0, 2048, dtype=torch.float32)
signal = torch.sin(2 * torch.pi * 440 / rate * sample)
waveform = signal.unsqueeze(0)
transform = torchaudio.transforms.MelSpectrogram(sample_rate=rate, n_mels=32)
mel = transform(waveform)
print('mel spectrogram:', mel.shape)
```
