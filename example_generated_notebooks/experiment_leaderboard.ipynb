# From experiments to a research leaderboard

Compare recorded validation histories with no bespoke Python analysis.

**Important:** These values are synthetic and illustrative, not

results from real training or evidence of model quality.

```python
import hy
hy.macros.require('lispyn.macros', None, target_module_name='experiment_leaderboard', assignments=[['experiment-leaderboard', 'experiment-leaderboard']], prefix='')
from lispyn.reproducibility import write_json
write_json('runs/leaderboard-demo/baseline/history.json', {'train_loss': [1.2, 0.9, 0.75], 'val_loss': [1.25, 0.95, 0.88]})
write_json('runs/leaderboard-demo/tuned/history.json', {'train_loss': [1.1, 0.8, 0.56], 'val_loss': [1.15, 0.81, 0.63]})
write_json('runs/leaderboard-demo/compact/history.json', {'train_loss': [1.3, 1.0, 0.82], 'val_loss': [1.38, 1.08, 0.92]})
from lispyn.leaderboard import show_leaderboard
board = show_leaderboard('runs/leaderboard-demo', metric='loss', split='val', rank_by='best', save='runs/leaderboard-demo/leaderboard.md')
print('Ranked synthetic histories:', board['total'])
```
